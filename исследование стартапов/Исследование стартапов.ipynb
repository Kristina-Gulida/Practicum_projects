Сегменты, к которым относится более 120 компаний, отнесите к массовым, сегменты, в которые входит от 35 до 120 включительно, отнесите к средним, а сегменты до 35 компаний отнесите к нишевым. Постройте график распределения количества компаний в сегментах и отобразите на нём разделение на нишевые и средние сегменты.

````python
segment_counts = pd.Series({
    'Массовые': mass_segments,
    'Средние': medium_segments,
    'Нишевые': niche_segments
})
segment_counts.plot(kind='bar', figsize=(8, 5))
plt.ylabel('Количество сегментов')
plt.title('Распределение сегментов рынка')
plt.xticks(rotation=0)

plt.show()
````

Дальнейшие исследования выполняйте с учётом этой замены. Индивидуальные сегменты внутри средней и нишевой групп рассматривать не нужно — они объединяются в два общих сегмента.

### 3.1. Анализируем и помечаем выбросы в каждом из сегментов

Заказчика интересует обычный для рассматриваемого периода размер средств, который предоставлялся компаниям.

По предобработанному столбцу `funding_total_usd` графическим образом оцените, какой размер общего финансирования для одной компании будет типичным, а какой — выбивающимся. Укажите интервал, в котором лежат типичные значения.

````python
limit = df['funding_total_usd'].quantile(0.99)
df[df['funding_total_usd'] <= limit]['funding_total_usd'].plot(
    kind='hist',
    bins=50,
    figsize=(8, 5)
)
plt.xlabel('Общий объём финансирования, USD')
plt.ylabel('Количество компаний')
plt.title('Распределение общего объёма финансирования')
plt.show()
````

Определите компании с аномальным объёмом общего финансирования — используйте метод IQR отдельно по каждому сегменту. Напомним, что все нишевые сегменты должны быть объединены в одну группу, а средние — в другую.

Определите сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведите топ таких сегментов.

````python
q1 = df.groupby('market')['funding_total_usd'].transform('quantile', 0.25)
q3 = df.groupby('market')['funding_total_usd'].transform('quantile', 0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
````

````python
outliers = df[
    (df['funding_total_usd'] < lower_bound) |
    (df['funding_total_usd'] > upper_bound)
]
outliers.head()
````

````python
outlier_share = (
    outliers['market'].value_counts()
    / df['market'].value_counts()
    * 100
)
outlier_share.sort_values(ascending=False).head(10)
````

````python
outlier_share.loc[['niche', 'mid']]
````

Для выявления аномального объёма финансирования применён метод межквартильного размаха (IQR) отдельно для каждого сегмента. Наибольшая доля компаний с аномальным финансированием наблюдается в сегментах Cloud Computing (14,38%), Technology (14,29%) и Big Data (14,29%). Для объединённых групп доля выбросов составила 10,50% для niche и 11,83% для mid.
